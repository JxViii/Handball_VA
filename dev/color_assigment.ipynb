```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
from sklearn.cluster import KMeans
```

```python
img_path = "../out/EON/crop_img.jpg"
img = cv2.imread(img_path)
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
```

```python
plt.imshow(img)
plt.show()
```

# Take top half of the player

```python
th_img = img[0: int(img.shape[0]/1.8), :]
plt.imshow(th_img)
plt.show()
```

# Cluster the image in two clusters

```python
# ! Reshape into 2d array

img_2d = th_img.reshape(-1,3)

# ! perform k-means clustering

kmeans = KMeans(n_clusters=2, random_state=0).fit(img_2d)

# ! get the labels

labels = kmeans.labels_

# ! reshape labels into og img shape

clustered_img = labels.reshape(th_img.shape[0], th_img.shape[1])

plt.imshow(clustered_img)
plt.show()

bg_color = 0
t_color = 0
```
